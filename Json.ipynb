```python
import requests 
from pprint import pprint

data = requests.get("http://api.open-notify.org/astros.json")

x = data.json()

pprint(x)

print('-----------------------------')
pprint(x['people'])
```

```python
import requests 
from pprint import pprint

data = requests.get("http://api.open-notify.org/astros.json")

x = data.json()



print('-----------------------------')
pprint(x['people'][0])

print('-----------------------------')

for each in x['people']:
    print(f"{each['name']}\t=========\t{each['craft']}")
```

```python
import json

mydict = {
    "Name":"Manjunathan",
    "City":("chennai","Mumbai","Kovai"),
    "Year":None
}

print(mydict,type(mydict))

mydata= json.dumps(mydict)

print(mydata,type(mydata))
```
