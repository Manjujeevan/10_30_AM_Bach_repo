```python
a = 1
while a<=5:   # row
    b = 1
    while b<=5:   # column
        print("*",end='')
        b += 1
    a += 1
    print()
```

```python
row = int(input("Enter a Row Number :"))
column = int(input("Enter a column Number :"))
symbol = input("Enter a symbol Number :")


a = 1
while a<=row:   # row
    b = 1
    while b<=column:   # column
        print(symbol,end='')
        b += 1
    a += 1
    print()
```
