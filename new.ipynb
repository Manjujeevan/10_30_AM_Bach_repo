```python
class Calci:
    def add(self,a,b):
        return a+b
    def sub(self,a,b):
        return a-b
    def mul(self,a,b):
        return a*b
    def div(self,a,b):
        return a/b

C = Calci()
print(C.add(20,5))
print(C.sub(20,5))
print(C.mul(20,5))
print(C.div(20,5))
```

```python
class Calci:
    def __init__(self,a,b):
        self.a=a
        self.b=b

    def add(self):
            return self.a + self.b
    def sub(self):
            return self.a - self.b


class Triple(Calci):
    def __init__(self,a,b,c):
        Calci.__init__(self,a,b)
        self.c = c

    def add(self):
        return self.a + self.b + self.c
    def sub(self):
        return self.a - self.b - self.c
    def mul(self):
        return self.a * self.b * self.c



c = Calci(30,5)  # Instant variabul

print(c.add())
print(c.sub())

print("------------------")
t = Triple(30,10,5)

print(t.add())
print(t.mul())
print(t.sub())
```
