### **Escape sequences**  
* \ - for newline in editor
* \\\ - prints \
* \’ – prints ‘
* \” – prints “
* \n – prints a new line
* \t – prints a tab


```python
print("hello \\world\\")
print("10\\100")
print('I\'am Python world')
print("Hello\"World\"")
print("Hello\nWorld") # \n=> New Line
print("Hello\tWorld") # \t => tab
```
