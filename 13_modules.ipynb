### **RegEx Functions**

```python
import re

with open("C:\\Users\\manju\\Desktop\\vengat.txt")as myfile:
    mydata = myfile.read()

    mylist = re.findall(r"[Aa][n][d]",mydata)
    print(mylist)
```

```python
import re

with open("C:\\Users\\manju\\Desktop\\vengat.txt")as myfile:
    mydata = myfile.read()

    mylist = re.findall(r"\d{2}[-/.]\d{2}[-/.]\d{4}",mydata)
    print(mylist)

    for each in mylist:
        print(each)
```

```python
# split()
import re

with open("C:\\Users\\manju\\Desktop\\vengat.txt")as myfile:
    mydata = myfile.read()

    mylist = re.split(r" ",mydata)

    for each in mylist:
        print(each)
  
 
  
```

```python
# sub()
import re

with open("C:\\Users\\manju\\Desktop\\vengat.txt")as myfile:
    mydata = myfile.read()

    mylist = re.sub(r" ",'\n',mydata)

    print(mylist)
```

```python
# search()
import re

with open("C:\\Users\\manju\\Desktop\\vengat.txt")as myfile:
    mydata = myfile.read()

    mylist = re.search(r"\d{2}[-/.]\d{2}[-/.]\d{4}",mydata)

    print(mylist)
```
