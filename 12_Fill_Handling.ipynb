### **Python File Handling**

```python
myfile = open("C:/Users/manju/Desktop/Mani.txt")

print(myfile.tell())

data = myfile.read(100)
print(data)
print(myfile.tell())

print('----------------')
data = myfile.read(200)
print(data)
print(myfile.tell())

myfile.seek(0)
print(myfile.tell())


print('----------------')
data = myfile.read(500)
print(data)
print(myfile.tell())
myfile.close()
```

**With Keyword**

```python
with open("C:\\Users\\manju\\Desktop\\Mani.txt")as myfile:

    myline = myfile.readline()
    print(myline)
    
```

```python
with open("C:\\Users\\manju\\Desktop\\Mani.txt")as myfile:

    mylines = myfile.readlines()
    print(mylines)
    print('---------------------------------')

    print(mylines[2])

    print('---------------------------------')

    for each in mylines:
        print(each)
```

### **"a" - Append mode**

```python
with open("C:\\Users\\manju\\Desktop\\Mani.txt", 'a')as myfile:
    print(myfile.tell())

    mydata = myfile.write("\nWelcome All")
```

### **"w" - Write - mode**

```python
with open("C:\\Users\\manju\\Desktop\\Mani.txt", 'w')as myfile:
    print(myfile.tell())

    mydata = myfile.write("Hello this GenAi class!!!!!!!!!")
```
