```python
a = 101

if a > 100:
    print(a,"is bigger then 100")

print("This line Alwas work")




```

```python
a = 100

if a > 100:
    print(a,"is bigger then 100")


elif a == 100:
    print(a,"is equal to 100")

else:
    print(a,"is smaller then 100")


print("This line Alwas work")
```

### **If ,elif, else rules.**  
* **if (Only once per code block(Mandtory))**
* **elif(as many time as per code block(optional))**
* **else(Only once per code black(optional))**

```python
a = 25

if a>50:
    print("a is bigger")

elif a == 20:
    print("a is 20")

elif a == 30:
    print("a is 30")

elif a == 40:
    print("a is 40")

else:
    print("a is smaller")
```

```python
# -----------------
# Game No one
# -----------------

score = 100
secret_number = 50
guess = int(input("Enter your guess: "))

if guess == secret_number:
    print("Correct!")
    score = score + 20
    print("Your Score:", score)

elif guess > secret_number:
    print("Too High!")
    score = score - 10
    print("Your Score:", score)

else:
    print("Too Low!")
    score = score - 10
    print("Your Score:", score)
```
