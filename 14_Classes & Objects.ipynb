### **Data Abstraction**


```python
class Clothing(object):
    def __init__(self, type, color, size, price=None):
        self.type = type
        self.color = color
        self.size = size
        self.price = price

    def set_price(self, price):
        # Set the price of an item of clothing.
        self.price = price
        print(f"Setting the price of the {self.color} {self.type} to ${price}.")

    def get_price(self):
        # Get the price of an item of clothing, if price is set.
        try:
            assert self.price != None
            print(f"The {self.color} {self.type} costs ${self.price}.")
        except:
            print(f"The price of the {self.color} {self.type} hasn't been set yet!")
    
    def promote(self, percentage):
        # Lower the price, if initial price is set.
        try:
            self.price = self.price * (1-percentage/100)
            print(f"The price of the {self.color} {self.type} has been reduced by {percentage} percent! It now only costs ${self.price:.0f}.")
        except:
            print(f"Oops. Set an initial price first!")

bluejeans = Clothing("jeans", "blue", 12)

bluejeans.get_price()
bluejeans.set_price(100)
bluejeans.get_price()
bluejeans.promote(25)
bluejeans.get_price() 
```

### **Inheritance**  
Inheritance allows us to define a class that inherits all the methods  and properties from another class. Parent class is the class being  inherited from, also called base class. Child class is the class that  inherits from another class, also called derived class.


```python
# Single inheritance

class A:
    a = 100

    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable are = {self.a}")

class B(A):
    b = 200

    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable are = {self.b}")

obj = B()
obj.myfunc()
print()
obj.newfunc()
```

```python
# Multiple inheritance: 

class A:
    a = 100

    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable are = {self.a}")

class B:
    b = 200

    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable are = {self.b}")

class C(A,B):
    c = 300

    def Oldfunc(self):
        print("This is Oldfunction Calling")
        print(f"This is C class variable are = {self.c}")


obj = C()

obj.myfunc()
print()
obj.newfunc()
print()
obj.Oldfunc()
```

```python
# Multilevel inheritance:

class A:
    a = 100

    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable are = {self.a}")

class B(A):
    b = 200

    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable are = {self.b}")

class C(B):
    c = 300

    def Oldfunc(self):
        print("This is Oldfunction Calling")
        print(f"This is C class variable are = {self.c}")

obj = C()

obj.myfunc()
print()
obj.newfunc()
print()
obj.Oldfunc()
```

```python
# Hierarchical inheritance

class A:
    a = 100

    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable are = {self.a}")

class B(A):
    b = 200

    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable are = {self.b}")

class C(A):
    c = 300

    def Oldfunc(self):
        print("This is Oldfunction Calling")
        print(f"This is C class variable are = {self.c}")


obj1 = B()
obj1.myfunc()
print()
obj1.newfunc()

print('--------------------')
obj2 = C()
obj2.myfunc()
print()
obj2.Oldfunc()
```

```python
# Hybrid inheritance 

class A:
    a = 100

    def myfunc(self):
        print("This is Myfunction Calling")
        print(f"This is A class variable are = {self.a}")

class B(A):
    b = 200

    def newfunc(self):
        print("This is Newfunction Calling")
        print(f"This is B class variable are = {self.b}")

class C(A):
    c = 300

    def Oldfunc(self):
        print("This is Oldfunction Calling")
        print(f"This is C class variable are = {self.c}")


class D(B,C):
    d = 400

    def yourfunc(self):
        print("This is Yourfunction Calling")
        print(f"This is D class variable are = {self.d}")


obj = D()
obj.myfunc()
print()
obj.newfunc()
print()
obj.Oldfunc()
print()
obj.yourfunc()
```
