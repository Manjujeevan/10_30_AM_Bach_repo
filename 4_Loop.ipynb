### **Python While loop** *(condition statements)*

*Syntax*  
* **while \<condition\>:**

```python
a = 1   # Initilization/Start 
while a<=10:    # Condition/End 
    print(a,end=" ")
    a += 1      #Incrementation/Jump
print("\nLoop Over")
```

**break** – *stop the loop*


```python

# Infinite Loop
a = 1

while a>0:
    print(a,end=" ")
    a += 1
```

```python
a = 1

while a>0:
    if a == 5:
        print(a)
        break
    print(a,end=" ")
    a += 1
```

**continue – go back to the top of my loop**


```python
a = 1
while a<=20:
    if a % 3 == 0:
        print(a,end=" ")
        a += 1
        continue
    
    a += 1

print("\nLoop over")
```

### **Python for Loop**  
Syntax

**for \<variable_name\> in \<str/list/tuple/set/dict\>:**

```python
mystr = "Hello world"

for x in mystr:   # Fetches Letter by letter
    print(x,end="")

```

```python
mylist = [10,20.55,"hello",True,20.55,"hello"]

for x in mylist:
    print(x,end =",")
```

**range()** – *is a python standard function used for numbers*


```python
for x in range(10+1): # end before
    print(x,end=' ')
```

```python
for x in range(5,10+1): # start with end before
    print(x,end=' ')
```

```python
num = int(input("enter a number :"))

for x in range(0,num+1):
    print(x,end=' ')
```

```python
for x in range(100,1000+1,100): # Jump
    print(x,end=" ")
```

```python
for x in range(100,0,-10): # reverse
    print(x,end=" ")
```

### **for with else**

```python
for x in "hello world":
    print(x,end='')
```

```python
# If Loop completes successfully, else will worke
# If loop is  unsuccessfully , else will not worke 

for x in "Hello world":
   print(x,end = "")
else:
    print("\nThe above loop complieted successfully")
```

```python
for x in "Helloworld":
   if x == ' ':
       break
   print(x,end = "")
else:
    print("\nThe above loop complieted successfully")
```

```python
# find if a number is divisible by 5 in the list

mylist = [2,6,9,10,15,11,16,97]

for each in mylist:
    if each % 5 == 0:
        print(each,"Divisible By 5")
        break
else:
    print("Number divisible by 5 not found")
```

### **Nested Loop**

```python
""" 
12345
12345
12345
12345
12345
 """

for x in range(5): # Row
    for y in range(1,5+1):
        print(y,end='')
    print()
```

```python
for x in ["Yellow","Green","Red"]:
    
    print("----------------------")
    for y in ["Cat","Dog","Car","House"]:
        print(x,y)
    print("----------------------")


print("Loop Over")
```
