```python
old_list = [1,2,3,4,5]

def square(x):
    return x*x

ans = list(map(square,old_list))
print(ans)
```

```python
old_list = [1,2,3,4,5]

def square(x):
    return x*x

def cube(n):
    return n*n*n

ans = list(map(square,old_list))
print(ans)

ans2 = list(map(cube,old_list))  # [1, 8, 27, 64, 125]
print(ans2)
```

### **Generator function**   
**yield** - keyword

```python
def myfunc():
    x = 50
    y =50
    z = x+y
    return z
    
ans = myfunc()
print(ans)
```

```python
def myfunc():
    yield 100
    yield "manju"
    yield 300.55

for i in myfunc():
    print(i,type(i))
```

```python
def myfunc():
    print("Hello Myfunc calling 1")
    yield  100
    print("Hello Myfunc calling 2")
    print("Hello Myfunc calling 3")
    yield 300
    print("Hello Myfunc calling 4")
    yield 400
    print("Hello Myfunc calling 5")
    yield 500
    

for each in myfunc():
    print("Loop in next iteration")
    print(each)
```

```python
x = 100
y = 200


print(f"X in Global scope = {x}")
print(f"Y in Global scope = {y}")
print()

def myfunc(x=0,y=0):
    print(f"X in Local scope = {x}")
    print(f"Y in Local scope = {y}")

myfunc()
print()
myfunc(1000,500)
myfunc()
```

### **Recursion**  

```python
def myadd(x,y):
    return x + y

def mysub(x,y):
    return x - y

#      15      30            15
print(mysub(myadd(10,20),myadd(5,10)))
```

```python
def myfactorial(num):
    pro = 1
    for i in range(1,num+1):
        pro = pro * i
    return pro
ans = myfactorial(5)
print(ans)
```

```python
def myrec_factorial(num):
    if num == 1:
        return 1
    else:
        return num * myrec_factorial(num - 1)
ans = myrec_factorial(5)
print(ans)
```

```python
#Fibonacci series using recursion

def recur_fibo(n):
   if n <= 1:
       return n
   else:
       return(recur_fibo(n-1) + recur_fibo(n-2))
nterms = 10
# check if the number of terms is valid
if nterms <= 0:
   print("Plese enter a positive integer")
else:
   print("Fibonacci sequence:")
   for i in range(nterms):
       print(recur_fibo(i))
```
