### **What is Nested Error Handling in Python?**

</center>

```python
try:
    print("Outer try block")

    try:
        num = int(input("Enter a number: "))
        result = 10 / num
        print(result)

    except ValueError:
        print("Inner Error: Invalid input (not a number)")

    except ZeroDivisionError:
        print("Inner Error: Cannot divide by zero")

except Exception as e:
    print("Outer Error:", e)
```

### **Task 1: List Index and Input Error**

*Create a program that:*

* Gets a list of numbers from the user.
* Asks the user to enter an index.
* Use nested try-except blocks to handle:
* ValueError
* IndexError

```python
try:
    # Get list of numbers from user
    numbers = list(map(int, input("Enter numbers separated by space: ").split()))

    try:
        # Get index from user
        index = int(input("Enter the index: "))

        try:
            print("Value:", numbers[index])

        except IndexError:
            print("Error: Index is out of range.")

    except ValueError:
        print("Error: Please enter a valid index.")

except ValueError:
    print("Error: Please enter only numbers in the list.")
```

### **Task 2: Dictionary Access**

*Create a student dictionary:*


student = {  
    "name": "Manju",  
    "age": 25,  
    "course": "Python"  
}   

Ask the user for a key and display its value.  

Use nested exception handling to handle:  

* Invalid input
* KeyError

```python
student = {
    "name": "Manju",
    "age": 40,
    "course": "Python"
}

try:
    key = input("Enter a key: ")

    if not key:
        raise ValueError("Key cannot be empty.")

    try:
        print("Value:", student[key])

    except KeyError:
        print("Error: Key not found.")

except ValueError as e:
    print("Error:", e)
```

### **Task 3: ATM Transaction**

**Create an ATM program with:**

* **balance = 10000**

Ask the user to enter withdrawal amount.

Use nested exception handling to handle:

* Non-numeric input → ValueError
* Withdrawal amount 0 → handle appropriately
* Withdrawal amount greater than balance → display insufficient balance
* Negative withdrawal amount → display invalid amount

*Expected output:*

* Enter withdrawal amount: 5000
* Withdrawal Successful
* Remaining Balance: 5000

```python
balance = 10000

try:
    amount = float(input("Enter withdrawal amount: "))

    try:
        if amount == 0:
            raise ValueError("Withdrawal amount cannot be zero.")

        if amount < 0:
            raise ValueError("Withdrawal amount cannot be negative.")

        if amount > balance:
            print("Insufficient balance.")

        else:
            balance = balance - amount
            print("Withdrawal Successful")
            print("Withdrawn Amount:", amount)
            print("Remaining Balance:", balance)

    except ValueError as e:
        print("Error:", e)

except ValueError:
    print("Error: Please enter a valid numeric amount.")
```
